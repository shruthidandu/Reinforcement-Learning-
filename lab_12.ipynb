```python

gamma = 0.9
theta = 0.001

states = ['S0', 'S1', 'G']
actions = ['Up', 'Down', 'Left', 'Right']

V = {
    'S0': 0,
    'S1': 0,
    'G': 0
}

transitions = {
    'S0': {
        'Up': 'S0',
        'Down': 'S1',
        'Left': 'S0',
        'Right': 'S0'
    },
    'S1': {
        'Up': 'S0',
        'Down': 'S1',
        'Left': 'S1',
        'Right': 'G'
    }
}

def get_reward(next_state):
    if next_state == 'G':
        return 100
    return -1

print("STATE, ACTION, REWARD AND TRANSITION INFORMATION")
print("-" * 55)

for state in ['S0', 'S1']:
    for action in actions:
        next_state = transitions[state][action]
        reward = get_reward(next_state)
        print(state, action, reward, next_state)

iteration = 0

print("\nITERATION-WISE STATE VALUES")
print("-" * 40)
print("Iteration\tV(S0)\tV(S1)")

while True:
    iteration += 1
    new_V = V.copy()

    for state in ['S0', 'S1']:
        q_values = []

        for action in actions:
            next_state = transitions[state][action]
            reward = get_reward(next_state)

            q = reward + gamma * V[next_state]
            q_values.append(q)

        new_V[state] = max(q_values)

    delta = max(abs(new_V[s] - V[s]) for s in ['S0', 'S1'])

    V = new_V

    print(iteration, "\t\t", round(V['S0'], 2),
          "\t", round(V['S1'], 2))

    if delta < theta:
        break

print("\nFINAL OPTIMAL STATE VALUES")
print("-" * 40)

for state in states:
    print("V(", state, ") =", V[state])

print("\nFINAL ACTION VALUE TABLE")
print("-" * 55)

policy = {}

for state in ['S0', 'S1']:
    q_values = {}

    for action in actions:
        next_state = transitions[state][action]
        reward = get_reward(next_state)

        q = reward + gamma * V[next_state]
        q_values[action] = q

        print(state, action, "Q =", round(q, 2))

    policy[state] = max(q_values, key=q_values.get)

print("\nOPTIMAL POLICY")
print("-" * 40)

for state in ['S0', 'S1']:
    print(state, "->", policy[state])

print("\nOPTIMAL PATH")
print("S0 -> S1 -> G")
```
