```python
# ============================================================
# ASSIGNMENT 1: UNDISTCOUNTED AND DISCOUNTED RETURNS
# ============================================================

print("===================================================")
print("ASSIGNMENT 1: RETURN CALCULATION")
print("===================================================")

# Accept number of time steps
n = int(input("Enter the number of time steps/episodes: "))

# Accept reward values
rewards = []

for i in range(n):
    reward = float(input(f"Enter reward for time step {i + 1}: "))
    rewards.append(reward)

# Calculate undiscounted return
undiscounted_return = sum(rewards)

# Accept discount factor
gamma = float(input("Enter the discount factor gamma (0 <= gamma <= 1): "))

# Validate discount factor
if gamma < 0 or gamma > 1:
    print("Invalid discount factor. Gamma must be between 0 and 1.")
else:
    # Calculate discounted return
    discounted_return = 0

    for t in range(n):
        discounted_return += (gamma ** t) * rewards[t]

    # Display results
    print("\n--------------- RESULTS ---------------")
    print("Entered reward values:", rewards)
    print("Undiscounted return:", undiscounted_return)
    print("Discounted return:", discounted_return)


# ============================================================
# ASSIGNMENT 2: 2 x 3 GRID WORLD MDP
# ============================================================

print("\n\n===================================================")
print("ASSIGNMENT 2: GRID WORLD MDP")
print("===================================================")

# ------------------------------------------------------------
# 1. STATE SPACE
# ------------------------------------------------------------

states = [
    "S0", "S1", "S2",
    "S3", "S4", "S5"
]

print("\nState Space:")
print(states)


# ------------------------------------------------------------
# 2. ACTION SPACE
# ------------------------------------------------------------

actions = ["UP", "DOWN", "LEFT", "RIGHT"]

print("\nAction Space:")
print(actions)


# ------------------------------------------------------------
# 3. SPECIAL STATES
# ------------------------------------------------------------

# S1 = obstacle
# S5 = goal

OBSTACLE = "S1"
GOAL = "S5"


# ------------------------------------------------------------
# 4. REWARD FUNCTION
# ------------------------------------------------------------

REWARD_NORMAL = -1
REWARD_GOAL = 10
REWARD_OBSTACLE = -5


# ------------------------------------------------------------
# 5. DISCOUNT FACTOR
# ------------------------------------------------------------

gamma_mdp = 0.9

print("\nDiscount Factor:")
print(gamma_mdp)


# ------------------------------------------------------------
# 6. GRID POSITIONS
# ------------------------------------------------------------

positions = {
    "S0": (0, 0),
    "S1": (0, 1),
    "S2": (0, 2),
    "S3": (1, 0),
    "S4": (1, 1),
    "S5": (1, 2)
}

# Reverse mapping
state_from_position = {
    position: state
    for state, position in positions.items()
}


# ------------------------------------------------------------
# 7. TRANSITION FUNCTION
# ------------------------------------------------------------

def transition(state, action):

    # Goal state remains in the goal state
    if state == GOAL:
        return GOAL

    row, col = positions[state]

    # Determine new position
    if action == "UP":
        new_row = row - 1
        new_col = col

    elif action == "DOWN":
        new_row = row + 1
        new_col = col

    elif action == "LEFT":
        new_row = row
        new_col = col - 1

    elif action == "RIGHT":
        new_row = row
        new_col = col + 1

    else:
        return state

    new_position = (new_row, new_col)

    # If outside grid, stay in current state
    if new_position not in state_from_position:
        return state

    next_state = state_from_position[new_position]

    # If obstacle, stay in current state
    if next_state == OBSTACLE:
        return state

    return next_state


# ------------------------------------------------------------
# 8. REWARD FUNCTION
# ------------------------------------------------------------

def get_reward(state, action, next_state):

    # Reaching goal
    if next_state == GOAL:
        return REWARD_GOAL

    # Check whether action attempted to enter obstacle
    row, col = positions[state]

    if action == "UP":
        new_position = (row - 1, col)

    elif action == "DOWN":
        new_position = (row + 1, col)

    elif action == "LEFT":
        new_position = (row, col - 1)

    elif action == "RIGHT":
        new_position = (row, col + 1)

    else:
        return REWARD_NORMAL

    if new_position in state_from_position:
        attempted_state = state_from_position[new_position]

        if attempted_state == OBSTACLE:
            return REWARD_OBSTACLE

    return REWARD_NORMAL


# ------------------------------------------------------------
# DISPLAY GRID
# ------------------------------------------------------------

print("\nGrid World:")
print("+-----+-----+-----+")
print("| S0  | S1* | S2  |")
print("+-----+-----+-----+")
print("| S3  | S4  | S5G |")
print("+-----+-----+-----+")
print("* = Obstacle")
print("G = Goal")


# ------------------------------------------------------------
# 9. GET CURRENT STATE AND ACTION
# ------------------------------------------------------------

current_state = input("\nEnter current state: ").upper()

if current_state not in states:
    print("Invalid state!")

else:

    action = input(
        "Enter action (UP/DOWN/LEFT/RIGHT): "
    ).upper()

    if action not in actions:
        print("Invalid action!")

    else:

        # Estimate next state using transition function
        next_state = transition(
            current_state,
            action
        )

        # Calculate reward
        reward = get_reward(
            current_state,
            action,
            next_state
        )

        # ----------------------------------------------------
        # DISPLAY TRANSITION RESULT
        # ----------------------------------------------------

        print("\n--------------- MDP RESULT ---------------")
        print("Current State :", current_state)
        print("Action        :", action)
        print("Next State    :", next_state)
        print("Reward        :", reward)
        print("Discount      :", gamma_mdp)

        print("\nTransition completed successfully.")
5
```
