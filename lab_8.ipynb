```python
gamma = 0.9

# Rewards obtained while following the policy
rewards = {
    "S0": 1,
    "S1": 2,
    "S3": 3,
    "S4": 4
}

policy = {
    "S0": "Right",
    "S1": "Down",
    "S3": "Right",
    "S4": "Up"
}


def state_value_function(state):
    if state == "S0":
        return rewards["S0"] + gamma * rewards["S1"] + gamma**2 * rewards["S3"] + gamma**3 * rewards["S4"]

    return 0


def action_value_function(state, action):
    if state == "S0" and action == "Up":
        return rewards["S0"]

    return 0


v_s0 = state_value_function("S0")
q_s0_up = action_value_function("S0", "Up")

print("Policy:", policy)
print()
print("State Value Function")
print("V(S0) =", v_s0)

print()
print("Action Value Function")
print("Q(S0, Up) =", q_s0_up)
```
