```python

gamma = 0.9
theta = 0.001

states = ["S0", "S1"]
actions = ["Up", "Down", "Left", "Right"]

rewards = {
    "normal": -1,
    "wall": -1,
    "goal": 10
}

goal = "G"
wall = "W"

transitions = {
    "S0": {
        "Up": ("S0", -1),
        "Down": ("S0", -1),
        "Left": ("S0", -1),
        "Right": ("S1", -1)
    },
    "S1": {
        "Up": ("S1", -1),
        "Down": ("G", 10),
        "Left": ("S0", -1),
        "Right": ("S1", -1)
    }
}

print("STATES:", states + ["WALL", "G"])
print("ACTIONS:", actions)
print("REWARDS:", rewards)
print("DISCOUNT FACTOR:", gamma)

print("\nTRANSITION MAPPING")
for state in states:
    for action in actions:
        next_state, reward = transitions[state][action]
        print(state, "--", action, "-->", next_state,
              "Reward:", reward)

V = {
    "S0": 0,
    "S1": 0,
    "G": 0
}

print("\nINITIAL STATE VALUES")
print(V)

print("\nVALUE ITERATION")

iteration = 0

while True:
    iteration += 1

    new_V = V.copy()

    for state in states:
        values = []

        for action in actions:
            next_state, reward = transitions[state][action]

            q = reward + gamma * V[next_state]
            values.append(q)

        new_V[state] = max(values)

    delta = max(abs(new_V[s] - V[s]) for s in states)

    V = new_V

    print("\nIteration:", iteration)
    print("S0 =", round(V["S0"], 4))
    print("S1 =", round(V["S1"], 4))
    print("Delta =", round(delta, 4))

    if delta < theta:
        break

print("\nFINAL OPTIMAL STATE VALUES")
for state in states:
    print(state, "=", round(V[state], 4))

print("\nACTION VALUE FUNCTION Q*(s,a)")

Q = {}

for state in states:
    Q[state] = {}

    for action in actions:
        next_state, reward = transitions[state][action]

        q = reward + gamma * V[next_state]
        Q[state][action] = q

        print(state, action, "=", round(q, 4))

print("\nOPTIMAL POLICY")

for state in states:
    max_q = max(Q[state].values())

    best_actions = [
        action for action in actions
        if abs(Q[state][action] - max_q) < 1e-9
    ]

    print(state, ":", best_actions)

print("\nBELLMAN OPTIMALITY VERIFICATION")

for state in states:
    max_q = max(Q[state].values())

    print(
        state,
        "V*(s) =", round(V[state], 4),
        "Max Q*(s,a) =", round(max_q, 4),
        "Verified =", abs(V[state] - max_q) < 1e-9
    )
```
