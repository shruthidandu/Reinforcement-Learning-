```python
import math

# Part A - Machine Initialization
machines = ['A', 'B', 'C']

M = 3
N_play = 10

# Part B - Initialize N, R, UCB and Q
N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}
UCB = {'A': 0.0, 'B': 0.0, 'C': 0.0}

total_reward = 0

# Reward patterns given in the assignment
reward_pattern = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

# Keep track of which reward to use next
reward_index = {
    'A': 0,
    'B': 0,
    'C': 0
}

# --------------------------------------------------
# Play the game for 10 plays
# --------------------------------------------------

for play in range(1, N_play + 1):

    print("\n========== Play", play, "==========")

    # Calculate UCB
    for machine in machines:

        if N[machine] == 0:
            UCB[machine] = float('inf')
        else:
            UCB[machine] = Q[machine] + math.sqrt(
                (2 * math.log(play)) / N[machine]
            )

    # Select machine with highest UCB
    selected = max(machines, key=lambda x: UCB[x])

    print("UCB Values:")

    for machine in machines:
        if UCB[machine] == float('inf'):
            print(machine, "= Infinity")
        else:
            print(machine, "=", round(UCB[machine], 4))

    print("Selected Machine:", selected)

    # --------------------------------------------------
    # Get reward from the given reward pattern
    # --------------------------------------------------

    index = reward_index[selected]

    reward = reward_pattern[selected][index]

    # Update reward_index using modulo to wrap around
    reward_index[selected] = (reward_index[selected] + 1) % len(reward_pattern[selected])

    print("Reward:", reward)

    # --------------------------------------------------
    # Update N
    # --------------------------------------------------

    N[selected] += 1

    # --------------------------------------------------
    # Update R
    # --------------------------------------------------

    R[selected] += reward

    # --------------------------------------------------
    # Incremental Sample Average
    # Qnew = Qold + (Reward - Qold) / N
    # --------------------------------------------------

    old_Q = Q[selected]

    Q[selected] = Q[selected] + (
        reward - Q[selected]
    ) / N[selected]

    total_reward += reward

    print("N =", N[selected])
    print("R =", R[selected])
    print("Q =", round(Q[selected], 4))


# --------------------------------------------------
# Final Results
# --------------------------------------------------

print("\n======================================")
print("       FINAL RESULTS")
print("======================================")

print("\nMachine\tN\tR\tUCB\t\tQ")

for machine in machines:

    if N[machine] == 0:
        UCB[machine] = float('inf')
        ucb_value = "Infinity"
    else:
        UCB[machine] = Q[machine] + math.sqrt(
            (2 * math.log(N_play)) / N[machine]
        )
        ucb_value = round(UCB[machine], 4)

    print(
        machine,
        "\t", N[machine],
        "\t", R[machine],
        "\t", ucb_value,
        "\t\t", round(Q[machine], 4)
    )

# --------------------------------------------------
# Total Reward
# --------------------------------------------------

print("\nTotal Reward =", total_reward)

# Best machine
best_machine = max(machines, key=lambda x: Q[x])

print("Best Machine =", best_machine)
print("Highest Q-value =", round(Q[best_machine], 4))
```
